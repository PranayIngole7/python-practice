### 1. Add Two Numbers
   
Take two numbers from the user and display their sum.

```python
num1 = 10
num2 = 20
sum = num1 + num2
print(f"Addition of {num1} and {num2} is {sum}.")
```

### 2. Subtract Two Numbers
Take two numbers and display the difference.

```python
num1 = 90
num2 = 20
diffn = num1 - num2
print(f"Subtraction of {num1} and {num2} is {diffn}.")
```

### 3. Multiply Two Numbers
Take two numbers and display their multiplication. 

```python
num1 = 9
num2 = 2
product = num1 * num2
print(f"Multiplication of {num1} and {num2} is {product}.")
```

### 4. Divide Two Numbers
Take two numbers and display the division result. 

```python
num1 = 50
num2 = 2
division = num1 / num2
print(f"Division of {num1} and {num2} is {division}.")
```

### 5. Calculate Square
Take a number and display its square. 

```python
num = 5
square = num * num
print(f"Square of {num} is {square}.")
```

### 6. Calculate Cube
Take a number and display its cube. 

```python
num = 5
cube = num * num * num
print(f"Cube of {num} is {cube}.")
```

### 7. Calculate Average of Two Numbers
Take two numbers and calculate their average.

```python
num1 = 50
num2 = 2
avg = (num1 + num2) / 2
print(f"Average of {num1} and {num2} is {avg}.")
```

### 8. Calculate Average of Three Numbers
Take three numbers and display their average.

```python
num1 = 50
num2 = 25
num3 = 75
avg = (num1 + num2 + num3) / 3
print(f"Average of {num1}, {num2} and {num3} is {avg}.")
```

### 9. Calculate Area of a Rectangle
Take length and width and calculate the area. 

```python
length = 13
width  = 7
area = length * width
print(f"The area of the rectangle is {area} square units.")
```

### 10. Calculate Perimeter of a Rectangle
Take length and width and calculate the perimeter. 

```python
length = 13
width  = 7
perimeter = 2 * (length + width)
print(f"The perimeter of the rectangle is {perimeter} units.")
```
